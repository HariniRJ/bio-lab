```python

import random

# Patients and available beds
patients = ["P1", "P2", "P3", "P4"]
beds = ["B1", "B2", "B3", "B4"]

# Patient priority: 1 = Normal, 2 = Medium, 3 = Emergency
priority = [1, 2, 3, 2]

# Allocation cost (waiting time)
cost = [
    [4, 5, 6, 7],
    [5, 3, 6, 4],
    [7, 6, 3, 5],
    [4, 6, 5, 3]
]

# CS parameters
n = 5
Pa = 0.25
max_iter = 50

# Fitness function
def fitness(solution):
    total = 0
    for i in range(4):
        total += cost[i][solution[i]]

        # Give preference to emergency patients
        total += (3 - priority[i]) * 2

    return total

# Generate initial nests
nests = []

for i in range(n):
    nests.append(random.sample(range(4), 4))

# Find best nest
best = min(nests, key=fitness)

# Main Cuckoo Search loop
for iteration in range(max_iter):

    for i in range(n):

        # Generate new solution (simplified Levy flight)
        new_solution = nests[i].copy()

        a, b = random.sample(range(4), 2)
        new_solution[a], new_solution[b] = (
            new_solution[b], new_solution[a]
        )

        # Evaluate fitness
        if fitness(new_solution) < fitness(nests[i]):
            nests[i] = new_solution

        # Update best solution
        if fitness(nests[i]) < fitness(best):
            best = nests[i].copy()

    # Abandon worst nests
    for i in range(n):
        if random.random() < Pa:
            nests[i] = random.sample(range(4), 4)

    # Update best solution
    current_best = min(nests, key=fitness)

    if fitness(current_best) < fitness(best):
        best = current_best.copy()

# Display result
print("Best Hospital Patient Allocation:")

for i in range(4):
    print(patients[i], "->", beds[best[i]])

print("Minimum Allocation Cost:", fitness(best))
print("Total Iterations:", max_iter)
```

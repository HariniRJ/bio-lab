```python

import random

# Patient priority: 1 = Normal, 2 = Medium, 3 = Emergency
patients = ["P1", "P2", "P3", "P4"]
beds = ["B1", "B2", "B3", "B4"]

priority = [1, 2, 3, 2]

# Allocation cost for each patient and bed
cost = [
    [4, 5, 6, 7],
    [5, 3, 6, 4],
    [7, 6, 3, 5],
    [4, 6, 5, 3]
]

# ACO parameters
ants = 10
alpha = 1
beta = 2
rho = 0.5
Q = 100
max_iter = 50

# Initial pheromone values
pheromone = [[1 for j in range(4)] for i in range(4)]

best_cost = float('inf')
best_solution = None

# ACO iterations
for iteration in range(max_iter):

    for ant in range(ants):

        assigned = []
        available = list(range(4))
        total_cost = 0

        # Allocate beds to patients
        for i in range(4):

            probabilities = []

            for j in available:
                heuristic = 1 / cost[i][j]
                value = (pheromone[i][j] ** alpha) * (heuristic ** beta)
                probabilities.append(value)

            # Select a bed based on probability
            total = sum(probabilities)
            probabilities = [p / total for p in probabilities]

            selected = random.choices(
                available, weights=probabilities, k=1
            )[0]

            assigned.append(selected)
            total_cost += cost[i][selected]

            available.remove(selected)

        # Check best allocation
        if total_cost < best_cost:
            best_cost = total_cost
            best_solution = assigned.copy()

        # Evaporate pheromones
        for i in range(4):
            for j in range(4):
                pheromone[i][j] *= (1 - rho)

        # Deposit pheromones on selected allocation
        for i in range(4):
            j = assigned[i]
            pheromone[i][j] += Q / total_cost

# Display result
print("Best Hospital Resource Allocation:")

for i in range(4):
    print(patients[i], "->", beds[best_solution[i]])

print("Minimum Allocation Cost:", best_cost)
print("Total Iterations:", max_iter)
```

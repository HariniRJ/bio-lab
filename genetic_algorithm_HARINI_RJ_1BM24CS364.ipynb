```python
import random

# Multi-Crop Irrigation Scheduling using Genetic Algorithm

# Crop details
crops = ["Rice", "Wheat", "Maize", "Cotton"]

# Water requirement of each crop (litres)
water_required = [500, 400, 300, 350]

# Benefit obtained per unit of water allocated
crop_benefit = [8, 7, 6, 5]

# Total available water in the region
available_water = 1000

# Genetic Algorithm parameters
POPULATION_SIZE = 20
GENERATIONS = 100
MUTATION_RATE = 0.1
CROSSOVER_RATE = 0.8


# Generate an initial population
def create_individual():
    return [
        random.randint(0, water_required[i])
        for i in range(len(crops))
    ]


def create_population():
    return [
        create_individual()
        for _ in range(POPULATION_SIZE)
    ]


# Fitness function
def fitness(individual):
    total_water = sum(individual)

    if total_water > available_water:
        return 0

    benefit = sum(
        individual[i] * crop_benefit[i]
        for i in range(len(crops))
    )

    return benefit


# Selection
def selection(population):
    population.sort(key=fitness, reverse=True)
    return population[:POPULATION_SIZE // 2]


# Crossover
def crossover(parent1, parent2):
    if random.random() < CROSSOVER_RATE:
        point = random.randint(1, len(crops) - 1)

        child = (
            parent1[:point] + parent2[point:]
        )
        return child

    return parent1.copy()


# Mutation
def mutation(individual):
    for i in range(len(individual)):
        if random.random() < MUTATION_RATE:
            individual[i] = random.randint(
                0, water_required[i]
            )

    return individual


# Main Genetic Algorithm
population = create_population()

for generation in range(GENERATIONS):

    # Evaluate and select the best individuals
    selected = selection(population)

    # Create the next generation
    new_population = selected.copy()

    while len(new_population) < POPULATION_SIZE:
        parent1 = random.choice(selected)
        parent2 = random.choice(selected)

        child = crossover(parent1, parent2)
        child = mutation(child)

        new_population.append(child)

    population = new_population

    # Best solution in the current generation
    best = max(population, key=fitness)

    if (generation + 1) % 10 == 0:
        print(
            f"Generation {generation + 1}: "
            f"Fitness = {fitness(best)}"
        )


# Display the final irrigation schedule
best = max(population, key=fitness)

print("\nOptimal Irrigation Schedule")
print("--------------------------------")

total_water = 0
total_benefit = 0

for i in range(len(crops)):
    allocation = best[i]
    benefit = allocation * crop_benefit[i]

    print(
        f"{crops[i]}: "
        f"Water = {allocation} litres, "
        f"Benefit = {benefit}"
    )

    total_water += allocation
    total_benefit += benefit

print("--------------------------------")
print("Total Water Used:", total_water, "litres")
print("Available Water:", available_water, "litres")
print("Total Crop Benefit:", total_benefit)
```
